### set methods
- adding element to set
  -add(ele):it adds element to set if element not exists in set , otherwise nothing
  -update(iterable):it adds all iteable elements to set
- remove elements from set
  - remove(ele):it removes the element from set if element exists in set otherwise it raises 'key error'
- discard(ele): it removes the elements from set if the element exists in set otherwise nothing
- pop():it removes the value from the set as well as it also return the value
- set operations
  -union(another_sets): it returns the all common elements from all sets
  -intersection(another_sets): it returns the common elements from all sets
  -difference(another_sets): it returns the set elements those are not lies in another_set
  -symmetric_difference(another_sets):it returns the non -common elements from all sets


```python
s={1,2,3,4}
s.add(7)
print(s)
s.update([10,20,30]) #it adds list elements
print(s)
s.remove(2)
print(s)
s.discard(5)
print(s)
s.pop()
print(s)
#s.remove(2)#key error
s1={1,2,3,4}
s2={3,4,5,6}
s3=(s1 | s2) #union
print(s3)
s4=(s1 & s2) #intersection
print(s4)
s5=(s1-s2) #difference
print(s5)
s6=(s1 ^ s2)#symmetric_difference
print(s6)
```
