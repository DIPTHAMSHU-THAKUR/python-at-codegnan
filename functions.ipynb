```python
#factorial 
def factorial(n):
    if n==1 or n==0:
        return 1
    return n*factorial(n-1)
print(factorial(4))
```

```python
#fibinoci 
def fun(n):
    if n==1 or n==0:
        return 1
    return fun(n-1) + fun(n-2)
print(fun(4))
```

```python
#sum of n natural number
def natur(n):
    if n==1:
        return 1
    return n + natur(n-1)
print(natur(5))
```

```python
# binary to decimal using recursion 
def  binar(n):
    if n==0:
        return "0"
    if n==1:
        return "1"
    return binar(n//2) + str(n%2)
print(binar(10))
```
