### list methods 
- element adding methods
  - apppend(item) : it adds item at end of the list
  - extend(itarable) : it adds all iterable elements at end of the list 
  - insert(pos, item) : this method  adds the item at specified position
    

```python
lst = [1,2,3,4]
lst.append(10) # we dont get the lst in the append statement it only append in the list to get the updated list we need to print again 
print(lst)
lst.append('code')
print(lst)
lst.append([10,30,20])
print(lst)
print(lst[6])
lst.extend([11,31,22])
print(lst)
lst.insert(1,2000)
print(lst)
```

###  element removing method
  - remove (val): it removes the first occurance of the value if value exists in the list , otherwise it raises 'value error'
  - pop(): it removes and returns the last of element of list
  - pop(pos):it removes and returns the specified position element from list 
  - clear(): it removes all element from list
  - del:  it python keyword , used to delete values


```python
lst =[10,20,30,20,40]
lst.remove(20)
print(lst)
#lst .remove(100) ValueError: list.remove(x): x not in list
val = lst.pop()
print(val,lst)
val2 = lst.pop(1)
print(val2, lst)
del lst[1]
print(lst)
lst.clear()
print(lst)
```
