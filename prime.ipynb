```python
num=int(input())
factors=[]
for i in range(1,num+1):
    if num%i==0:
        factors.append(i)
print(factors)
if len(factors)==2:
    print(f"{num} is prime")
else:
    print(f"{num} is not prime")
    
```

```python
s=input()
rev=""
for char in s:
    rev=char + rev
print(rev)
if rev==s:
    print("it is palindrome")
else:
    print("not a palindrome")
```
