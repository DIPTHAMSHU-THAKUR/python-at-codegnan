##### string methods
- case conversion methods
  - upper() it converts all letters to upper 
  - lower() it converts all letters to lower 
  - title()it converts the first letter in every word to upper  and non first to lower case
  - swapcase() it converts upper to lower or if in lower it converts to upper
  - capitalize()it converts first letter to upper  and remaining to lower case

```python
string="the ocean is BLue"
lower= string.lower()
print(lower)
```

```python
s="the boy iS Playing"
print(s.lower())
print(s.upper())
print(s.title())
print(s.swapcase())
print(s.capitalize())

```

```python
# white space remove methods 
# strip():it removes the both side white space  
# lstrip():it removes the left side space 
# rstrip(): it removes the right side space 

s ="              codegnan                "
print(s)
print(s.strip())
print(s.lstrip())
print(s.rstrip())

```

```python
#split():it returns the substring and by default splits at whitespace 
s="the boy is very good@"
print(s.split())
print(s.split('is'))
print(s.split('z'))
print(s.split('good'))
print(s.split('@'))

```

```python
# indexing finding methods 
# index(sub_string) it gives the index number of the first ocurance and the start place of the letter and if not found it gives value error 
# find(sub_string)  it gives the index number of the first ocurance and the start place of the letter and if not found it gives -1
s="The ocean is blue and sky is blue"
print(s.find("o"))
print(s.index("is"))
print(s.index("is",14))#when we add a number it starts checking the index from 14
print(s.find("z"))
```

```python
# replace (old,new): it replaces the old_string with new_string 
s="The ocean is blue and sky is blue"
print(s.replace('blue','red'))

```

```python
# "sep". join(iterable): it joins the all iterable elements with seperators 
lst= ['the', 'boy', 'is', 'very', 'Good']
print(" ".join(lst))
print("   ".join(lst))
print("_".join(lst))
print(" b  ".join(lst))


```

#### string checking methods 
- islower(): it returns true when entire string is in  lower case
- isupper(): it returns true when entire string is in upper case
- istitle(): it returns true when string is in title format
- isalpha(): it returns true when string contains alphabets
- isalnum(): it returns true when string contains either  alphabets or  numbers
- isdigit(): it returns true when string  contains  only digits 
- startswith(substring): it returns true when string starts with substring 
- endswith(substring): it returns true when string ends with substring


```python
s="theboyis8good"
print(s.islower())
print(s.isupper())
print(s.istitle())
print(s.isalpha())
print(s.isalnum())
print(s.isdigit())
print(s.startswith('t'))
print(s.endswith('d'))


```

```python
# count(sub_string) : it returns count  of non overlaping sub_string occurance 
print("ABCCBCA".count('C'))
print("ABCBCB".count('BCB'))
```
