```python
#generator defination 
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res =gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
print(next(res))
```

```python
#generator defination 
#to overcome error using loops
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res =gen(5)
for i in res:
    print(i)
```

```python
def fun(n):
    res=[]
    for i in range(1,n+1):
        res.append(i)
    return res
n=int(input())
print(fun(n))
```

```python
def gen(n):
    for i in range(1,n+1):
        yield i
n=gen(int(input()))
for i in n:
    print(i)
```
