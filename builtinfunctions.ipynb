# built in functions  definations is provided by the python , those functions called as built in functions

```python
#abs(): It returns the absoulte value 
print(abs(20-10))
print(abs(10-20))
```

```python
# round(val,pos): this function  rounds  the float value to particular  positions 
print(round (2.2589,2))
print(round (2.2521,2))
print(round (2.2589))
print(int(2.2589))
```

```python
#ord(cahr): it returns the ACSII value of character
#chr(asci_val): it returns the character for specified ASCII value 
print(ord("A"))
print(chr(65))
```

```python
#sorted(iterable,reverse= False , key =None ): it returns new sorted iterable 
# sort the friuits based on the length of fruit name 
fruits=['Apple','banana','kivi','Mango', 'Water Melon']
res = sorted(fruits)
print(res)
print(sorted(fruits,reverse=True))
print(sorted(fruits,reverse=True,key=len))
print(sorted(fruits,reverse=False,key=len))# it sort the lists based on length of each  element length or item length 

```

```python
#map(function , iterable)
# find the square of all list of integers using map function 
def square(x):
     return x**x
arr=[1,2,3,4,5]
v=list(map(square,arr))
print(v)



    
```

```python
#filter(function , iterable )
# find even numbers from list of numbers
def check_even(x):
    if x%2==0:
        return x
lst=[1,2,3,4,5]
res= list(filter(check_even, lst))
print(res)
```

```python
# find the 3's multipliers from list of  integers 
def mul_ti(x):
    if x%3==0:
        return x
arr=[1,24,54,22,44,51,33,15]
res=list(filter(mul_ti,arr))
print(res)
```

```python
# enumerate(iterable): this returns the list of tuple of index and item for iterable items 
fruits=['Apple','banana','kivi','Mango', 'Water Melon']
print(enumerate(fruits))
print(list(enumerate(fruits)))
for ind, val in enumerate(fruits):
    print(f"{ind}--{val} and length is {len(val)}")
```

```python
#tuple unpacking
t=(1,2)
a,b=t
print(a,b)
```

```python
#zip(iterable1,iterable2,...): it returns all itables as  single  iteable of tples 
rool_no=[101,102,103,104,105]
names=['Ravi','Raju','siva',"Rani","Gopi"]
percentage=[90,67,96.8,67.65,99.9999]
print(list(zip(rool_no,names,percentage)))
```

```python
#any() : this function returns True when any one iterable items return True 
#all(): this fuction returns True when all iterable items return True
arr=[1,2,3,4,0]
print(all(arr))
print(any(arr))
```

```python
percentage =[34,35,54,65,67]
for num in percentage:
    if num<35:
        print("all students not passed")
        break
    else:
        print("all students  passed")
```

```python
percentage =[34,35,54,65,67]
if all(percentage)>35:
    print("pass")
else:
    print("not all pass")
```

```python
percentage =[34,35,54,65,67]
lst=[]
for num in percentage:
    if num<35:
        lst.append(False)
    else:
        lst.append(True)
print(lst)
if all(lst):
    print("all students pass")
else:
    print("not all pass")
```
